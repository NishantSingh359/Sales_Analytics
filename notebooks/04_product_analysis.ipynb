### Category Level Metrics

```python
ca_rev = pkpi.category_revenue()
ca_pro = pkpi.category_profit()
ca_unit = pkpi.category_units()

vs.ColumnPlot(
    x=ca_rev["category"],
    y1=ca_rev["revenue"],
    title="Revenue by Product Category",
    ylabel="Profit",
    xlabel="Category",
).decoration()

vs.ColumnPlot(
    x=ca_pro["category"],
    y1=ca_pro["profit"],
    ylabel="Profit",
    xlabel="Category",
    title="Profit by Product Category",
).decoration()

vs.ColumnPlot(
    x=ca_unit["category"],
    y1=ca_unit["quantity"],
    ylabel="Quantity",
    xlabel="Category",
    title="Quantity by Product Category",
).decoration()
```

### Product Level Metrics

#### Revenue Per Product

```python
df = pkpi.revenue_per_product().head(50)
vs.LinePlot(
    x=df["product_name"],
    y1=df["revenue"],
    title="Top Product by Revenue",
    xlabel="Product",
    ylabel="Revenue",
    fsize=(15, 6),
).decoration()
```

#### Total Sold Quantity Per Product

```python
df = pkpi.units_sold_per_product().head(50)
vs.LinePlot(
    x=df["product_name"],
    y1=df["quantity"],
    title="Product Sales Volume",
    xlabel="Product",
    ylabel="Quantity",
    fsize=(15, 6),
).decoration()
```

#### Total Profit Per Product

```python
df = pkpi.profit_per_product().head(50)
vs.LinePlot(
    x=df["product_name"],
    y1=df["profit"],
    title="Total Profit Per Product",
    xlabel="Product",
    ylabel="Profit",
    fsize=(20, 8),
).decoration()
```

#### Average Product Price

```python
df = pkpi.avg_product_price().head(50)
vs.LinePlot(
    x=df["product_name"],
    y1=df["price"],
    title="Average Product Price",
    xlabel="Product",
    ylabel="Average Price",
    fsize=(20, 8),
).decoration()
```
